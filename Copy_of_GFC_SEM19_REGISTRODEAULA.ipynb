```python
# Instalar a biblioteca necessária
!pip install openpyxl

from openpyxl import Workbook
from google.colab import files
import os
import base64
from IPython.display import HTML, display

# Criar a planilha
wb = Workbook()
ws = wb.active
ws.title = "AP2"

# Cabeçalho
ws["A1"] = "Lista de transações em dinheiro"
ws["D1"] = "PASSO 2"
ws["D2"] = "Elaborar o fluxo de caixa pelo método direto"

# Dados principais
dados = [
    ("Venda de produtos", 1500000),
    ("Recebimento de empréstimo", 250000),
    ("Pagamento de fornecedores", 70000),
    ("Salários", 175000),
    ("Compra de máquina", 250000),
    ("Juros e encargos financeiros", 850000),
    ("Pagamento de terceiros", 50000),
    ("Impostos e taxas", 125000),
]

for i, (desc, valor) in enumerate(dados, start=2):
    ws[f"A{i}"] = desc
    ws[f"B{i}"] = valor

# ===== Fluxo de caixa operacional =====
ws["D3"] = "Fluxo de caixa operacional"
ws["E3"] = 230000

ws["D4"] = "Entradas"
ws["E4"] = 1500000

ws["D5"] = "Saídas"
ws["E5"] = 1270000

# Saídas operacionais detalhadas
ws["D6"] = "Detalhamento das saídas:"
ws["D7"] = "Pagamento de fornecedores"
ws["E7"] = 70000
ws["D8"] = "Salários"
ws["E8"] = 175000
ws["D9"] = "Juros e encargos financeiros"
ws["E9"] = 850000
ws["D10"] = "Pagamento de terceiros"
ws["E10"] = 50000
ws["D11"] = "Impostos e taxas"
ws["E11"] = 125000

# ===== Fluxo de caixa de investimento =====
ws["D13"] = "Fluxo de caixa de investimento"
ws["E13"] = -250000

ws["D14"] = "Saídas"
ws["E14"] = 250000
ws["D15"] = "Compra de máquina"
ws["E15"] = 250000

# ===== Fluxo de caixa de financiamento =====
ws["D17"] = "Fluxo de caixa de financiamento"
ws["E17"] = 250000

ws["D18"] = "Entradas"
ws["E18"] = 250000
ws["D19"] = "Recebimento de empréstimo"
ws["E19"] = 250000

ws["D20"] = "Saídas"
ws["E20"] = 0

# ===== Saldo final =====
ws["D22"] = "Saldo final"
ws["E22"] = 230000

# ===== PASSO 1 - Classificação =====
ws["A11"] = "PASSO 1"
ws["A12"] = "Classificar as transações por categoria"

classificacao = [
    ("Venda de produtos", "Operacional (entrada)"),
    ("Recebimento de empréstimo", "Financiamento (entrada)"),
    ("Pagamento de fornecedores", "Operacional (saída)"),
    ("Salários", "Operacional (saída)"),
    ("Compra de máquina", "Investimento (saída)"),
    ("Juros e encargos financeiros", "Operacional (saída)"),
    ("Pagamento de terceiros", "Operacional (saída)"),
    ("Impostos e taxas", "Operacional (saída)"),
]

for i, (desc, cat) in enumerate(classificacao, start=13):
    ws[f"A{i}"] = desc
    ws[f"B{i}"] = cat

# ===== Formatação =====
from openpyxl.styles import Font
bold_font = Font(bold=True)

for cell in ['A1', 'D1', 'D2', 'D3', 'D4', 'D5', 'D13', 'D14', 'D17', 'D18', 'D22', 'A11', 'A12']:
    ws[cell].font = bold_font

for col in ['A', 'B', 'D', 'E']:
    ws.column_dimensions[col].width = 30

# ===== Salvar =====
nome_arquivo = "solucao_fluxo_caixa.xlsx"
wb.save(nome_arquivo)

print(f"✅ Planilha '{nome_arquivo}' criada com sucesso!")

# ===== OPÇÃO 1: Download via files.download() =====
try:
    files.download(nome_arquivo)
    print("✅ Download iniciado via files.download()")
except Exception as e:
    print(f"⚠️ Erro no download automático: {e}")
    print("Usando método alternativo...")

# ===== OPÇÃO 2: Download via HTML (mais confiável) =====
def download_arquivo(arquivo):
    """Função para download via HTML"""
    with open(arquivo, "rb") as f:
        data = f.read()
    b64 = base64.b64encode(data).decode()
    href = f'<a href="data:application/vnd.openxmlformats-officedocument.spreadsheetml.sheet;base64,{b64}" download="{arquivo}">📥 <b>Clique aqui para baixar o arquivo</b></a>'
    display(HTML(href))
    print("✅ Clique no link acima para baixar o arquivo!")

download_arquivo(nome_arquivo)

# ===== OPÇÃO 3: Salvar no Google Drive (opcional) =====
try:
    from google.colab import drive
    drive.mount('/content/drive')

    # Salvar no Drive
    drive_path = f'/content/drive/MyDrive/{nome_arquivo}'
    wb.save(drive_path)
    print(f"✅ Arquivo também salvo no Google Drive: {drive_path}")
except:
    print("ℹ️ Para salvar no Google Drive, execute: drive.mount('/content/drive')")

# ===== RESUMO =====
print("\n" + "="*60)
print("📊 RESUMO DA CLASSIFICAÇÃO")
print("="*60)
print("\n📌 FLUXO OPERACIONAL:")
print("  Entrada: Venda de produtos (R$ 1.500.000)")
print("  Saídas:")
print("    - Pagamento de fornecedores: R$ 70.000")
print("    - Salários: R$ 175.000")
print("    - Juros e encargos financeiros: R$ 850.000")
print("    - Pagamento de terceiros: R$ 50.000")
print("    - Impostos e taxas: R$ 125.000")
print(f"  Total de saídas: R$ 1.270.000")
print(f"  ✅ Fluxo operacional líquido: R$ 230.000")

print("\n📌 FLUXO DE INVESTIMENTO:")
print("  Saída: Compra de máquina (R$ 250.000)")
print(f"  ✅ Fluxo de investimento líquido: -R$ 250.000")

print("\n📌 FLUXO DE FINANCIAMENTO:")
print("  Entrada: Recebimento de empréstimo (R$ 250.000)")
print(f"  ✅ Fluxo de financiamento líquido: R$ 250.000")

print("\n" + "="*60)
print("✅ VERIFICAÇÃO DO SALDO FINAL:")
print("  Saldo inicial: R$ 0")
print("  + Fluxo operacional: R$ 230.000")
print("  + Fluxo de investimento: -R$ 250.000")
print("  + Fluxo de financiamento: R$ 250.000")
print("  = Saldo final: R$ 230.000")
print("="*60)
print("✅ O saldo final do fluxo decomposto é IGUAL ao consolidado!")
print("="*60)

# Limpeza (opcional)
# os.remove(nome_arquivo)
```
